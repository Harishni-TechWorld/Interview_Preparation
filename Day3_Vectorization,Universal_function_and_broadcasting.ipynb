### Problem 1: Feature Standardization (Z-Score Normalization)
In Machine Learning, before feeding data into a neural network, you standardize every feature (column) so that each column has a **mean of 0** and a **standard deviation of 1**.

**Formula:**
$$Z = \frac{X - \mu}{\sigma}$$

**Starter Data:**
```python
import numpy as np

# 4 samples (rows), 3 features (columns: e.g., Age, Salary, Experience)
X = np.array([[22, 30000, 1],
              [25, 45000, 2],
              [30, 60000, 5],
              [35, 80000, 8]], dtype=float)

````python
import numpy as np

# 4 samples (rows), 3 features (columns: e.g., Age, Salary, Experience)
X = np.array([[22, 30000, 1],
              [25, 45000, 2],
              [30, 60000, 5],
              [35, 80000, 8]], dtype=float)
column_means=np.mean(X,axis=0)
print(column_means)
column_std=np.std(X,axis=0)
print(column_std)
Y=(X-column_means)/column_std
print(Y)
print("New Column Means:",Y.mean(axis=0))
print("New Column standard:",Y.std(axis=0))
````

### Problem 2: Pairwise Euclidean Distance Matrix (3D Broadcasting)
In clustering algorithms (like K-Means) and recommendation systems, you need the Euclidean distance between **every pair of points** in a dataset—without using nested `for` loops.

**Formula:**
$$\text{distance} = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}$$

**Starter Data:**
```python
import numpy as np

# 3 points in 2D space (x, y coordinates)
points = np.array([[0, 0],
                   [3, 4],
                   [6, 8]], dtype=float)

````python
import numpy as np

# 3 points in 2D space: (x, y) coordinates
points = np.array([[0, 0],
                   [3, 4],
                   [6, 8]], dtype=float)
p1=points.reshape(3,1,2)
p2=points.reshape(1,3,2)
diff=p1-p2
print(diff)

c1=diff**2
print(c1)
sum_sqrt=np.sum(c1,axis=-1)
dist_matrix=np.sqrt(sum_sqrt)
print(dist_matrix)

````

### Problem 3: Softmax Probability Layer (Ufuncs + Broadcasting)
Every multi-class neural network classifier ends with a **Softmax** layer that turns raw model scores (logits) into probabilities that sum to `1.0` across each row.

**Formula:**
$$\text{Softmax}(x_i) = \frac{e^{x_i}}{\sum_{j} e^{x_j}}$$

**Starter Data:**
```python
import numpy as np

# Raw scores for 2 images across 3 classes (Cat, Dog, Bird)
logits = np.array([[2.0, 1.0, 0.1],
                   [0.5, 2.5, 1.0]])

````python
import numpy as np

# Raw scores for 2 images across 3 classes (Cat, Dog, Bird)
logits = np.array([[2.0, 1.0, 0.1],
                   [0.5, 2.5, 1.0]])

exp_scores=np.exp(logits)
print(exp_scores)

sum=np.sum(exp_scores,axis=1,keepdims=True)
print(sum)
probs=(exp_scores/sum)
print(probs)
probs.sum(axis=1)
````

### Problem 4: Row-Wise L2 Normalization (Unit Vectors)

**Goal:** Given a 2D array `embeddings` of shape `(4, 3)`, scale each row independently so that its Euclidean length (L2 norm) equals `1.0`.

**Formula:**
$$\hat{x}_i = \frac{x_i}{\sqrt{\sum_{j} x_{i,j}^2}}$$

**Constraints:**
- Zero explicit `for` loops (pure NumPy vectorization and broadcasting).

**Verification Check:**
- `np.sum(normalized ** 2, axis=1)` should output `[1. 1. 1. 1.]`.


````python
import numpy as np

array=np.arange(12).reshape(4,3)
print(array)
normalize=array**2
print(normalize)

row_sum=np.sum(normalize,axis=1,keepdims=True)
print(row_sum)
#formula
l2_norm=np.sqrt(row_sum)
probs=(array/l2_norm)
print(probs)
print(np.sum(probs**2,axis=1))
````
