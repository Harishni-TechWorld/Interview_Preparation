```python
import numpy as np
list=([10,20,],[30,40],[50,60])
array=np.array(list,dtype='U32')

print(array)
```

```python
import numpy as np
array=np.zeros((3,4))

print(array.size)
print(array.shape)
```

```python
import numpy as np
array=np.arange(11,19).reshape((2,4))

print(array)
```

```python
import numpy as np
array_1=np.array([[10,20,30],[50,60,70]])

print(array_1.ndim)
print(array_1.shape)
print(array_1.size)
print(array_1.dtype)
print(array_1.itemsize)
```
